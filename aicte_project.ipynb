```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier
from sklearn.multioutput import MultiOutputClassifier

# Load your data
df = pd.read_csv("/nutrition_data.csv")

# Select features
X = df[["Calories", "Protein", "Carbs", "Fat", "Iron", "Calcium", "VitaminC",
        "VitaminD", "Fiber", "Sodium", "Activity", "Age"]]

# Define targets
y_def = df[["Low_Iron", "Low_VitaminD", "Low_Fiber"]]  # Multi-label classification
y_risk = df["At_Risk_Of_Disease"]  # Binary classification

# Split all together with same seed
X_train, X_test, y_def_train, y_def_test, y_risk_train, y_risk_test = train_test_split(
    X, y_def, y_risk, test_size=0.2, random_state=42
)
```

```python
model_def = MultiOutputClassifier(RandomForestClassifier(random_state=42))
model_def.fit(X_train, y_def_train)
pred_def = model_def.predict(X_test)
```

```python
model_risk = LogisticRegression()
model_risk.fit(X_train, y_risk_train)
pred_risk = model_risk.predict(X_test)
```

```python
print("X_train shape:", X_train.shape)
print("y_risk_train shape:", y_risk_train.shape)
```

```python
from sklearn.metrics import classification_report

print("Nutrient Deficiency Prediction Report:")
print(classification_report(y_def_test, pred_def, target_names=y_def.columns))
```

```python
from sklearn.metrics import classification_report

print(classification_report(y_risk_test, pred_risk, target_names=["Low Risk", "At Risk"], zero_division=0))
```

```python
def generate_tips(def_row, risk_pred):
    tips = []

    if def_row[0]:  # Low Iron
        tips.append("Increase iron intake: spinach, red meat, lentils.")
    if def_row[1]:  # Low Vitamin D
        tips.append("Get more sunlight and consume fortified foods.")
    if def_row[2]:  # Low Fiber
        tips.append("Eat more fruits, vegetables, and whole grains.")
    if risk_pred == 1:
        tips.append("You're at risk of chronic disease. Reduce sodium and sugar, increase physical activity.")

    return tips
```

```python
import pandas as pd

def generate_tips(def_row, risk_pred):
    tips = []

    if def_row[0]:  # Low Iron
        tips.append("Increase iron intake: spinach, red meat, lentils.")
    if def_row[1]:  # Low Vitamin D
        tips.append("Get more sunlight and consume fortified foods.")
    if def_row[2]:  # Low Fiber
        tips.append("Eat more fruits, vegetables, and whole grains.")
    if risk_pred == 1:
        tips.append("You're at risk of chronic disease. Reduce sodium and sugar, increase physical activity.")

    return tips


def predict_user_health(model_def, model_risk, input_dict):
    # Convert user input into DataFrame with feature names
    input_data = pd.DataFrame([input_dict])

    # Predict nutrient deficiencies and disease risk
    predicted_def = model_def.predict(input_data)
    predicted_risk = model_risk.predict(input_data)

    # Generate tips
    tips = generate_tips(predicted_def[0], predicted_risk[0])

    # Display results
    print("📊 Predicted Nutrient Deficiencies:")
    print(f"- Low Iron: {'Yes' if predicted_def[0][0] else 'No'}")
    print(f"- Low Vitamin D: {'Yes' if predicted_def[0][1] else 'No'}")
    print(f"- Low Fiber: {'Yes' if predicted_def[0][2] else 'No'}")
    print(f"\n⚠️ Chronic Disease Risk: {'At Risk' if predicted_risk[0] else 'Low Risk'}")

    print("\n💡 Personalized Recommendations:")
    for tip in tips:
        print(f"- {tip}")
```

```python
# Sample user input
user_input = {
    "Calories": 2000,
    "Protein": 80,
    "Carbs": 220,
    "Fat": 70,
    "Iron": 6,
    "Calcium": 600,
    "VitaminC": 50,
    "VitaminD": 150,
    "Fiber": 18,
    "Sodium": 2500,
    "Activity": 1,
    "Age": 30
}

# Call the function with models and input
predict_user_health(model_def, model_risk, user_input)
```
